```python
import pandas as pd

data = {
    "Student_ID": [101, 102, 103, 101, 105],
    "Midterm_Score": [85.0, np.nan, 72.0, 85.0, 45.0],
    "Study_Hours": [12.0, 6.0, np.nan, 12.0, 3.0],
    "City": ["Delhi", "Mumbai", "Delhi", "Delhi", "Chennai"],
    "Passed": ["Yes", "No", "Yes", "Yes", "No"]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

df = df.drop_duplicates()

numeric_cols = ["Midterm_Score", "Study_Hours"]

df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

df = pd.get_dummies(df, columns=["City"])

print("\nProcessed DataFrame:")
print(df)

X = df.drop("Passed", axis=1)
y = df["Passed"]

print("\nFeature Matrix (X):")
print(X)

print("\nTarget Vector (y):")
print(y)


# Question - 11
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42
)

print("X_train shape:", X_train.shape)
print("X_test shape :", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape :", y_test.shape)
```
