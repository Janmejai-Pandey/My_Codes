```python
import math

Doc_1 = "gamers gaming on gaming keyboards"
Doc_2 = "wireless mechanical keyboards for typing"
Doc_3 = "fast wireless charging pads"
Doc_4 = "ergonomic keyboards help gamers type fast"

docs = [Doc_1, Doc_2, Doc_3, Doc_4]

def tokenize(doc):
    tokens = doc.lower().split()
    
    stop_words = ["on", "for", "the", "help"]
    tokens = [token for token in tokens if token not in stop_words]
    
    return tokens

def lemmatize(tokens):
    lemme = {
        "gaming": "game",
        "gamers": "game",
        "keyboards": "keyboard",
        "pads": "pad",
        "typing": "Type",
        "type": "Type"
    }
    return [lemme.get(token, token) for token in tokens]

def TF(tokenized_docs):
    tf = {}
    
    for doc_id, tokenized_doc in enumerate(tokenized_docs, start=1):
        doc_tf = {}
        total_tokens = len(tokenized_doc)
        
        for token in tokenized_doc:
            doc_tf[token] = doc_tf.get(token, 0) + 1
            
        for token in doc_tf:
            doc_tf[token] /= total_tokens
            
        tf[doc_id] = doc_tf
        
    return tf

def IDF(tokenized_docs):
    idf = {}
    total_docs = len(tokenized_docs)
    
    for tokenized_doc in tokenized_docs:
        unique_tokens = set(tokenized_doc)
        for token in unique_tokens:
            idf[token] = idf.get(token, 0) + 1
            
    for token in idf:
        idf[token] = math.log(total_docs / idf[token])
        
    return idf

def TF_IDF(tf, idf):
    tf_idf = {}
    
    for doc_id, doc_tf in tf.items():
        doc_tf_idf = {}
        for token, tf_value in doc_tf.items():
            doc_tf_idf[token] = tf_value * idf.get(token)
        tf_idf[doc_id] = doc_tf_idf
        
    return tf_idf

def query(query_str, idf, tokenized_docs):
    query_tokens = tokenize(query_str)
    query_tokens = lemmatize(query_tokens)
    
    ranked_docs = {}
    
    for doc_id, doc in enumerate(tokenized_docs, start=1):
        ranked_docs[f"Doc_{doc_id}"] = 0
        for token in query_tokens:
            if token in doc:
                ranked_docs[f"Doc_{doc_id}"] += idf.get(token)
    
    ranked_docs = sorted(ranked_docs.items(), key=lambda x: x[1], reverse=True)
    return ranked_docs

if __name__ == "__main__":
    tokenized_docs = [lemmatize(tokenize(doc)) for doc in docs]
    
    tf = TF(tokenized_docs)
    idf = IDF(tokenized_docs)
    tf_idf = TF_IDF(tf, idf)
    
    query_str = input("Enter query - ")
    ranked_docs = query(query_str, idf, tokenized_docs)
    
    
    print("\nRanked Documents for Query:", query_str)
    for rank, (doc, score) in enumerate(ranked_docs, start=1):
        if score > 0:
            print(f"Rank {rank}: Document: {doc}, Score: {score}")

```
