```python
import math
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')
stop_words = set(stopwords.words('english'))

D1 = "Information retrieval systems help users find relevant documents from large collections of text."
D2 = "Search engines use indexing and ranking techniques to retrieve relevant information efficiently."
D3 = "Machine learning methods can improve document ranking and search relevance."
D4 = "Inverted indexes are fundamental data structures used in information retrieval systems."
D5 = "Natural language processing helps search engines understand user queries better."
D6 = "Ranking algorithms assign relevance scores to documents based on query terms."
D7 = "Search systems often combine indexing, retrieval, and ranking for effective information access."
D8 = "Deep learning approaches are increasingly used to improve retrieval performance."
D9 = "Information retrieval focuses on finding relevant information from document collections."
D10 = "Query processing and document indexing are important components of modern search engines."

docs = [D1, D2, D3, D4, D5, D6, D7, D8, D9, D10]

def tokenize(docs):
    tokenized_docs = []
    
    for doc in docs:
        tokens = doc.lower().split()
        tokens = [token for token in tokens if token not in stop_words]
        tokenized_docs.append(tokens)
        
    return tokenized_docs

def TF(tokenized_docs):
    tf = {}
    
    for doc_id, tokenized_doc in enumerate(tokenized_docs, start=1):
        doc_tf = {}
        total_tokens = len(tokenized_doc)
        
        for token in tokenized_doc:
            doc_tf[token] = doc_tf.get(token, 0) + 1
            
        for token in doc_tf:
            doc_tf[token] /= total_tokens
            
        tf[doc_id] = doc_tf
        
    return tf

def IDF(tokenized_docs):
    idf = {}
    total_docs = len(tokenized_docs)
    
    for tokenized_doc in tokenized_docs:
        unique_tokens = set(tokenized_doc)
        for token in unique_tokens:
            idf[token] = idf.get(token, 0) + 1
            
    for token in idf:
        idf[token] = math.log(total_docs / idf[token])
        
    return idf

def TF_IDF(tf, idf):
    tf_idf = {}
    
    for doc_id, doc_tf in tf.items():
        doc_tf_idf = {}
        for token, tf_value in doc_tf.items():
            doc_tf_idf[token] = tf_value * idf.get(token, 0)
        tf_idf[doc_id] = doc_tf_idf
        
    return tf_idf

def rank_documents(query, tf_idf):
    query_tokens = query.lower().split()
    scores = {}
    
    for doc_id, doc_tf_idf in tf_idf.items():
        for token in query_tokens:
            if token in doc_tf_idf:
                score = scores.get(doc_id, 0) + doc_tf_idf[token]
        scores[doc_id] = score
        
    ranked_docs = sorted(scores.items(), key=lambda item: item[1], reverse=True)
    
    return ranked_docs

if __name__ == "__main__":
    tokenized_docs = tokenize(docs)
    tf = TF(tokenized_docs)
    idf = IDF(tokenized_docs)
    tf_idf = TF_IDF(tf, idf)

    query = input("Enter your search query: ")
    ranked_docs = rank_documents(query, tf_idf)

    print("\nRanked Documents:")
    for doc_id, score in ranked_docs:
        print(f"Document {doc_id}: Score = {score:.4f}")
```
